```python
import plotly.express as px
import pandas as pd
import os

# 1. SETUP: Country list
data = [
    # --- EU 27 ---
    {'country': 'Austria', 'iso_alpha': 'AUT', 'status': 'MMTIS NAP usage'},
    {'country': 'Belgium', 'iso_alpha': 'BEL', 'status': 'MMTIS NAP usage'},
    {'country': 'Bulgaria', 'iso_alpha': 'BGR', 'status': 'No NAP usage yet'},
    {'country': 'Croatia', 'iso_alpha': 'HRV', 'status': 'No NAP usage yet'},
    {'country': 'Cyprus', 'iso_alpha': 'CYP', 'status': 'No NAP usage yet'},
    {'country': 'Czech Rep.', 'iso_alpha': 'CZE', 'status': 'No NAP usage yet'},
    {'country': 'Denmark', 'iso_alpha': 'DNK', 'status': 'MMTIS NAP usage'},
    {'country': 'Estonia', 'iso_alpha': 'EST', 'status': 'RTTI / SRTI NAP usage'},
    {'country': 'Finland', 'iso_alpha': 'FIN', 'status': 'RTTI / SRTI NAP usage'},
    {'country': 'France', 'iso_alpha': 'FRA', 'status': 'MMTIS NAP usage'},
    {'country': 'Germany', 'iso_alpha': 'DEU', 'status': 'MMTIS + RTTI / SRTI NAP usage'},
    {'country': 'Greece', 'iso_alpha': 'GRC', 'status': 'No NAP usage yet'},
    {'country': 'Hungary', 'iso_alpha': 'HUN', 'status': 'No NAP usage yet'},
    {'country': 'Ireland', 'iso_alpha': 'IRL', 'status': 'No NAP usage yet'},
    {'country': 'Italy', 'iso_alpha': 'ITA', 'status': 'No NAP usage yet'},
    {'country': 'Latvia', 'iso_alpha': 'LVA', 'status': 'No NAP usage yet'},
    {'country': 'Lithuania', 'iso_alpha': 'LTU', 'status': 'No NAP usage yet'},
    {'country': 'Luxembourg', 'iso_alpha': 'LUX', 'status': 'No NAP usage yet'},
    {'country': 'Malta', 'iso_alpha': 'MLT', 'status': 'No NAP usage yet'},
    {'country': 'Netherlands', 'iso_alpha': 'NLD', 'status': 'MMTIS + RTTI / SRTI NAP usage'},
    {'country': 'Poland', 'iso_alpha': 'POL', 'status': 'No NAP usage yet'},
    {'country': 'Portugal', 'iso_alpha': 'PRT', 'status': 'No NAP usage yet'},
    {'country': 'Romania', 'iso_alpha': 'ROU', 'status': 'No NAP usage yet'},
    {'country': 'Slovakia', 'iso_alpha': 'SVK', 'status': 'No NAP usage yet'},
    {'country': 'Slovenia', 'iso_alpha': 'SVN', 'status': 'RTTI / SRTI NAP usage'},
    {'country': 'Spain', 'iso_alpha': 'ESP', 'status': 'RTTI / SRTI NAP usage'},
    {'country': 'Sweden', 'iso_alpha': 'SWE', 'status': 'MMTIS NAP usage'},

    # --- NON-EU Specific ---
    {'country': 'Norway', 'iso_alpha': 'NOR', 'status': 'MMTIS + RTTI / SRTI NAP usage'},
    {'country': 'Iceland', 'iso_alpha': 'ISL', 'status': 'No NAP usage yet'},
    {'country': 'United Kingdom', 'iso_alpha': 'GBR', 'status': 'MMTIS NAP usage'},
    {'country': 'Switzerland', 'iso_alpha': 'CHE', 'status': 'MMTIS NAP usage'},
    {'country': 'Moldova', 'iso_alpha': 'MDA', 'status': 'No NAP usage yet'},
    {'country': 'Montenegro', 'iso_alpha': 'MNE', 'status': 'No NAP usage yet'},
]

df = pd.DataFrame(data)

# 2. COLORS (Google Brand Colors)
color_map = {
    'MMTIS NAP usage': '#0F9D58',
    'RTTI / SRTI NAP usage':  '#4285F4',
    'MMTIS + RTTI / SRTI NAP usage':  '#F4B400',
    'No NAP usage yet':  '#E0E0E0'
}

# 3. CREATE MAP
fig = px.choropleth(
    df,
    locations="iso_alpha",
    color="status",
    hover_name="country",
    color_discrete_map=color_map,
    projection="mercator", # Keeps upright shapes
)

# 4. CUSTOMIZE LAYOUT FOR "BUBBLE / STICKER" LOOK
fig.update_traces(
    marker_line_color='white', # The border color
    marker_line_width=3        # Very thick borders to create the "gap" effect
)

fig.update_geos(
    visible=False,
    resolution=110,        # KEY CHANGE: Low resolution = Smoother, less detailed lines
    showcountries=False,
    showcoastlines=False,
    showland=False,
    fitbounds=False,

    # Camera Crop (Adjusted to include Norway top)
    lataxis_range=[40, 74],
    lonaxis_range=[-25, 34]
)

fig.update_layout(
    width=800,
    height=900,
    margin={"r":0,"t":0,"l":0,"b":0},
    showlegend=False,
    dragmode=False,
    paper_bgcolor='rgba(255,255,255,1)', # White background for PNG
    plot_bgcolor='rgba(0,0,0,0)'
)

# 5. SHOW AND EXPORT
fig.show()

# This saves the file to your Colab folder
fig.write_image("google_EU_NAP_usage_map.png", scale=3)
print("Export complete! Please check the file browser on the left.")
```
